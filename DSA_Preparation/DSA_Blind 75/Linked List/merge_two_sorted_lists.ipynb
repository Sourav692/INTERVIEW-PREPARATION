# 21. Merge Two Sorted Lists
**Difficulty:** 🟢 Easy · **Topic:** Linked List · **LeetCode:** https://leetcode.com/problems/merge-two-sorted-lists/

## 💡 Concepts

**Core concept(s):** Weave two sorted lists together using a **dummy** node.

**Why it applies here:** Like merging two sorted piles of cards: repeatedly take the smaller front card. A dummy head lets you append without special-casing the first pick.

**Key intuition:** Always attach the smaller of the two front nodes, then advance that list.

---

### 📚 What is a Linked List?
A **linked list** is a chain of nodes; each node holds a value and a pointer to the **next** node. Unlike an array there is no index — you can only walk forward from the head.
- **In Python:** a small `ListNode` class with `.val` and `.next`.

### 📚 Pointers & the Dummy Node
Linked-list code moves **pointers** (references to nodes). A **dummy** node placed before the head removes annoying "is this the first node?" special cases — you build off `dummy.next` and return it at the end.

### 📚 Recursion on a List
A list problem often splits into "handle this node, then recurse on the rest". Clean to write, but uses stack depth equal to the list length.

---

**Prerequisite knowledge:**
- Dummy node.
- Pointer splicing.

## 📝 Problem

Merge two sorted lists into one sorted list.

**Example**
```
1->2->4  &  1->3->4  ->  1->1->2->3->4->4
```

> Two approaches, both `O(n+m)`: iterative with a dummy, and recursive.

````python
from typing import Optional, List

class ListNode:
    """A node in a singly linked list: a value plus a link to the next node."""
    def __init__(self, val=0, next=None):
        self.val = val                     # the value stored at this node
        self.next = next                   # link to the next node (None at the end)

def build_list(vals):
    """Turn a Python list into a linked list; return its head."""
    dummy = ListNode(); cur = dummy        # dummy node avoids special-casing the first node
    for v in vals:
        cur.next = ListNode(v); cur = cur.next
    return dummy.next

def to_list(head):
    """Turn a linked list back into a Python list (handy for printing / assertions)."""
    out = []
    while head:
        out.append(head.val); head = head.next
    return out
````

### Approach 1 — Iterative with Dummy (optimal)

**Idea:** Build off a dummy; attach the smaller front node each step; append whatever remains.

**Time:** `O(n+m)`. **Space:** `O(1)`.

````python
def merge_two_iter(a: Optional[ListNode], b: Optional[ListNode]) -> Optional[ListNode]:
    dummy = ListNode(); tail = dummy       # dummy head; tail is where we attach the next node
    while a and b:                         # while both lists still have nodes
        if a.val <= b.val:                 # attach the smaller front node
            tail.next = a; a = a.next
        else:
            tail.next = b; b = b.next
        tail = tail.next                   # advance the tail of the merged list
    tail.next = a or b                     # one list is empty; attach the rest of the other
    return dummy.next
````

### Approach 2 — Recursive

**Idea:** The smaller head, followed by the merge of the rest.

**Time:** `O(n+m)`. **Space:** `O(n+m)` stack.

````python
def merge_two_rec(a: Optional[ListNode], b: Optional[ListNode]) -> Optional[ListNode]:
    if not a: return b                     # a is empty -> the answer is the rest of b
    if not b: return a                     # b is empty -> the answer is the rest of a
    if a.val <= b.val:
        a.next = merge_two_rec(a.next, b)  # a's head is smaller; merge the rest after it
        return a
    else:
        b.next = merge_two_rec(a, b.next)  # b's head is smaller
        return b
````

````python
# Correctness check
tests = [([1,2,4],[1,3,4],[1,1,2,3,4,4]), ([],[0],[0]), ([],[],[])]
for a, b, exp in tests:
    assert to_list(merge_two_iter(build_list(a), build_list(b))) == exp
    assert to_list(merge_two_rec(build_list(a), build_list(b))) == exp
    print(a, b, "->", exp)
print("\nAll tests passed")
````

## ⏱️ Empirically Checking the Complexities

We time each approach on inputs of growing size `n` and read the **doubling ratio**.

| Theoretical | Ratio when `n` → `2n` |
|---|---|
| `O(n)`       | ≈ **2×** |
| `O(n log n)` | ≈ **2×** (slightly more) |
| `O(n²)`     | ≈ **4×** |

*(Recursive is O(n) too but stack-limited; we time the iterative merge.)*

````python
import os, sys
_root = os.getcwd()
for _ in range(5):
    if os.path.exists(os.path.join(_root, "bench_utils.py")): break
    _root = os.path.dirname(_root)
if _root not in sys.path: sys.path.insert(0, _root)
from bench_utils import benchmark

def make_worst_case(n):
    return (build_list(list(range(0, 2*n, 2))), build_list(list(range(1, 2*n, 2))))
solutions = {
    "iterative O(n+m)": merge_two_iter,
}
sizes = [20000, 40000, 80000, 160000]

benchmark(solutions, make_worst_case, sizes, plot=True)
````
