A **graph** = **vertices** (things) + **edges** (connections).

## 1. A graph as an adjacency list
One class handles both **undirected** and **directed** graphs.

```python
from collections import defaultdict

class Graph:
    """A simple graph stored as an adjacency list: vertex -> set of neighbours."""
    def __init__(self, directed=False):
        self.directed = directed
        self.adj = defaultdict(set)

    def add_edge(self, u, v):
        self.adj[u].add(v)
        if not self.directed:         # undirected -> the link goes both ways
            self.adj[v].add(u)

    def neighbours(self, u):
        return sorted(self.adj[u])

    def vertices(self):
        return sorted(self.adj)

g = Graph(directed=False)
for u, v in [("A","B"), ("A","D"), ("B","C"), ("B","E"), ("C","F"), ("D","E"), ("E","F")]:
    g.add_edge(u, v)
print("vertices      :", g.vertices())
print("A's neighbours:", g.neighbours("A"))
print("B's neighbours:", g.neighbours("B"))
```

## 2. Degree (undirected)
How many edges touch a vertex.

```python
def degree(g, u):
    """Undirected degree = number of neighbours."""
    return len(g.adj[u])

for v in g.vertices():
    print(f"  deg({v}) = {degree(g, v)}")
assert degree(g, "B") == 3
```

## 3. Directed graphs: in-degree vs out-degree

```python
dg = Graph(directed=True)
for u, v in [("X","Y"), ("Y","Z"), ("Z","X")]:    # a directed 3-cycle
    dg.add_edge(u, v)

def out_degree(g, u):
    """Arrows leaving u."""
    return len(g.adj[u])

def in_degree(g, u):
    """Arrows entering u = how many vertices list u as a neighbour."""
    return sum(1 for src in g.adj if u in g.adj[src])

for v in dg.vertices():
    print(f"  {v}: out={out_degree(dg, v)}  in={in_degree(dg, v)}")
assert out_degree(dg, "X") == 1 and in_degree(dg, "X") == 1
```

## 4. Reachability — does a path exist?
A DFS with a **visited set** (graphs have cycles, so we must not loop).

```python
def path_exists(g, start, goal):
    """Is `goal` reachable from `start`? DFS with a visited set."""
    seen, stack = set(), [start]
    while stack:
        u = stack.pop()
        if u == goal:
            return True
        if u in seen:                 # already explored -> skip (avoids cycles)
            continue
        seen.add(u)
        stack.extend(g.adj[u])
    return False

print("A -> F path?", path_exists(g, "A", "F"))

g2 = Graph()
g2.add_edge("P", "Q")
_ = g2.adj["Z"]                       # touch Z so it exists but stays isolated
print("P -> Z path?", path_exists(g2, "P", "Z"))
assert path_exists(g, "A", "F")
assert not path_exists(g2, "P", "Z")
```

## ✅ Recap
- Graph = **vertices + edges**; store as an **adjacency list** (`vertex -> neighbours`).
- **Undirected** adds both directions; **directed** keeps arrows (in-degree vs out-degree).
- Reachability / cycles need a **visited set** — the thing trees never required.
- Types (weighted, DAG, dense/sparse, bipartite) are in the written tutorial.
