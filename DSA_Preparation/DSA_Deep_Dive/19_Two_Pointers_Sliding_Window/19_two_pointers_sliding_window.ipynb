## 1. Two pointers on a sorted array -- opposite ends converging

```python
def two_sum_sorted(arr, target):
    """arr is SORTED. Find two indices whose values sum to target."""
    lo, hi = 0, len(arr) - 1
    while lo < hi:
        s = arr[lo] + arr[hi]
        if s == target:
            return [lo, hi]
        elif s < target:
            lo += 1
        else:
            hi -= 1
    return []

arr = [1, 2, 4, 7, 11, 15]
result = two_sum_sorted(arr, 9)
print("indices summing to 9:", result, "->", arr[result[0]], "+", arr[result[1]])
assert result == [1, 3]
assert two_sum_sorted(arr, 100) == []             # no pair sums to 100


def max_water_container(heights):
    """Container With Most Water: move the SHORTER wall inward -- it can't be part of a better answer."""
    lo, hi, best = 0, len(heights) - 1, 0
    while lo < hi:
        width = hi - lo
        area = width * min(heights[lo], heights[hi])
        best = max(best, area)
        if heights[lo] < heights[hi]:
            lo += 1
        else:
            hi -= 1
    return best

heights = [1, 8, 6, 2, 5, 4, 8, 3, 7]
result = max_water_container(heights)
print("max water container:", result)
assert result == 49    # between index 1 (height 8) and index 8 (height 7): width 7 * min(8,7) = 49
```

## 2. Sliding window -- variable-size, both pointers move only right

```python
def longest_substring_without_repeat(s):
    seen = {}
    left = best = 0
    for right, ch in enumerate(s):
        if ch in seen and seen[ch] >= left:
            left = seen[ch] + 1                    # shrink past the previous occurrence
        seen[ch] = right
        best = max(best, right - left + 1)
    return best

print("longest unique substring of 'abcabcbb':", longest_substring_without_repeat("abcabcbb"))
assert longest_substring_without_repeat("abcabcbb") == 3     # "abc"
assert longest_substring_without_repeat("bbbbb") == 1         # "b"
assert longest_substring_without_repeat("") == 0
assert longest_substring_without_repeat("pwwkew") == 3         # "wke"
```

## 3. Fast & slow pointers -- cycle detection

```python
class ListNode:
    def __init__(self, val):
        self.val = val
        self.next = None

def has_cycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            return True
    return False

# Build a cyclic list: 1 -> 2 -> 3 -> 2 (cycle back to node 2)
n1, n2, n3 = ListNode(1), ListNode(2), ListNode(3)
n1.next, n2.next, n3.next = n2, n3, n2
assert has_cycle(n1) is True

# A plain, acyclic list: 1 -> 2 -> 3 -> None
m1, m2, m3 = ListNode(1), ListNode(2), ListNode(3)
m1.next, m2.next = m2, m3
assert has_cycle(m1) is False
print("cycle detection: cyclic list ->", has_cycle(n1), ", acyclic list ->", has_cycle(m1))
```

## 4. Event / sweep-line -- two pointers generalized to intervals

```python
def max_overlap(intervals):
    events = []
    for start, end in intervals:
        events.append((start, 1))
        events.append((end, -1))
    events.sort(key=lambda e: (e[0], -e[1]))       # starts before ends at the same timestamp

    active = best = 0
    for _, delta in events:
        active += delta
        best = max(best, active)
    return best

example = [(2, 7), (4, 8), (15, 20)]
result = max_overlap(example)
print("max overlap:", result)
assert result == 2       # (4,8) and (2,7) overlap on [4,7)

triple = [(1, 10), (2, 9), (3, 8)]
assert max_overlap(triple) == 3        # all three overlap on [3,8)

disjoint = [(1, 2), (5, 6)]
assert max_overlap(disjoint) == 1       # nothing overlaps
```

## ✅ Recap
- **Opposite-ends two pointers**: needs sorted input; move the pointer on the "wrong" side inward -- O(n).
- **Sliding window**: both pointers move only right; expand then shrink to maintain a valid window -- O(n) total movement.
- **Fast/slow pointers**: different speeds through a linked structure -- detects cycles without extra memory.
- **Event/sweep-line**: generalizes two pointers to intervals -- convert to +1/-1 events, sort, sweep -- O(n log n).
- All four replace what would otherwise be an O(n^2) nested loop or an O(n * range) brute-force scan.
