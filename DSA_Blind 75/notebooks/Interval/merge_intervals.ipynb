# 56. Merge Intervals
**Difficulty:** 🟡 Medium · **Topic:** Interval · **LeetCode:** https://leetcode.com/problems/merge-intervals/

## 💡 Concepts

**Core concept(s):** **Sort by start**, then sweep merging any interval that overlaps the last kept one.

**Why it applies here:** Once sorted by start, overlapping intervals are adjacent. Keep a running 'current' interval; if the next one starts before current ends, stretch current's end; otherwise start a new one.

**Key intuition:** Sort by start; extend the current interval while the next one overlaps, else begin a new one.

---

### 📚 Sorting to Reveal Order
Sorting intervals by their start (or end) puts overlapping ones next to each other, so a single left-to-right pass can merge or count them. Cost: **O(n log n)**.

---

**Prerequisite knowledge:**
- Sorting.
- Overlap + merge.

## 📝 Problem

Merge all overlapping intervals.

**Example**
```
[[1,3],[2,6],[8,10],[15,18]] -> [[1,6],[8,10],[15,18]]
```

> One approach: sort + sweep, `O(n log n)`.

### Approach — Sort + Sweep

**Idea:** Sort by start. For each interval, either extend the last merged interval's end (overlap) or append a new one.

**Time:** `O(n log n)`. **Space:** `O(n)`.

````python
def merge_intervals(intervals):
    intervals.sort(key=lambda x: x[0])     # sort by start so overlaps become adjacent
    merged = []
    for start, end in intervals:
        if merged and start <= merged[-1][1]:   # this interval overlaps the current block
            merged[-1][1] = max(merged[-1][1], end)  # extend the block's end
        else:
            merged.append([start, end])    # no overlap -> start a new block
    return merged
````

````python
# Correctness check
tests = [
    ([[1,3],[2,6],[8,10],[15,18]],[[1,6],[8,10],[15,18]]),
    ([[1,4],[4,5]],[[1,5]]),
    ([[1,4],[2,3]],[[1,4]]),
]
for iv, exp in tests:
    got = merge_intervals([x[:] for x in iv])
    print(f"{iv} -> {got}")
    assert got == exp, "mismatch!"
print("\nAll tests passed")
````

We time each approach on inputs of growing size `n` and read the **doubling ratio**.

| Theoretical | Ratio `n`→`2n` |
|---|---|
| `O(n)`       | ≈ **2×** |
| `O(n log n)` | ≈ **2×** (slightly more) |
| `O(n²)`     | ≈ **4×** |
