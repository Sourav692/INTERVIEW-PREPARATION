# 57. Insert Interval
**Difficulty:** 🟡 Medium · **Topic:** Interval · **LeetCode:** https://leetcode.com/problems/insert-interval/

## 💡 Concepts

**Core concept(s):** One left-to-right pass over already-sorted intervals, merging any that overlap the new one.

**Why it applies here:** The list is sorted and non-overlapping. Copy intervals that end before the new one starts, merge every interval that touches the new one into it, then copy the rest.

**Key intuition:** Add the ones before, swallow the overlapping ones into the new interval, then add the ones after.

---

### 📚 Sorting to Reveal Order
Sorting intervals by their start (or end) puts overlapping ones next to each other, so a single left-to-right pass can merge or count them. Cost: **O(n log n)**.

---

**Prerequisite knowledge:**
- Interval overlap test.
- A three-phase scan.

## 📝 Problem

Insert a new interval into a sorted, non-overlapping list, merging as needed.

**Example**
```
[[1,3],[6,9]], new=[2,5] -> [[1,5],[6,9]]
```

> One approach: single `O(n)` pass (input is already sorted).

### Approach — Three-Phase Scan

**Idea:** Add intervals ending before `new` starts; merge all that overlap `new`; add the rest.

**Time:** `O(n)`. **Space:** `O(n)`.

````python
def insert_interval(intervals, new):
    res = []
    i, n = 0, len(intervals)
    while i < n and intervals[i][1] < new[0]:   # phase 1: intervals entirely BEFORE new
        res.append(intervals[i]); i += 1
    while i < n and intervals[i][0] <= new[1]:  # phase 2: intervals that OVERLAP new -> merge
        new = [min(new[0], intervals[i][0]), max(new[1], intervals[i][1])]  # widen new
        i += 1
    res.append(new)                        # add the merged interval
    while i < n:                           # phase 3: everything AFTER new
        res.append(intervals[i]); i += 1
    return res
````

````python
# Correctness check
tests = [
    ([[1,3],[6,9]],[2,5],[[1,5],[6,9]]),
    ([[1,2],[3,5],[6,7],[8,10],[12,16]],[4,8],[[1,2],[3,10],[12,16]]),
    ([],[5,7],[[5,7]]),
]
for iv, new, exp in tests:
    got = insert_interval([x[:] for x in iv], new[:])
    print(f"{iv} + {new} -> {got}")
    assert got == exp, "mismatch!"
print("\nAll tests passed")
````

## 🧩 Patterns Learned

- **Three-phase interval scan:** before / overlapping / after — clean handling of a merge-in.
- **Overlap test:** two intervals touch when `a.start <= b.end` and `b.start <= a.end`.
- **Signal:** "insert / add into sorted intervals".
- **Related problems:** Merge Intervals, Interval List Intersections.
- **Common pitfalls:** (1) strict vs non-strict overlap; (2) forgetting to append the merged interval.
