# 435. Non-overlapping Intervals
**Difficulty:** 🟡 Medium · **Topic:** Interval · **LeetCode:** https://leetcode.com/problems/non-overlapping-intervals/

## 💡 Concepts

**Core concept(s):** **Greedy** — sort by **end** time and always keep the interval that finishes earliest.

**Why it applies here:** To remove the fewest intervals, keep as many as possible. The interval that ends earliest leaves the most room for the rest, so keep it and drop anything that overlaps it. This classic 'activity selection' greedy is provably optimal.

**Key intuition:** Keep the earliest-finishing interval; drop whatever overlaps it; repeat.

---

### 📚 Sorting to Reveal Order
Sorting intervals by their start (or end) puts overlapping ones next to each other, so a single left-to-right pass can merge or count them. Cost: **O(n log n)**.

### 📚 What is Greedy?
**Greedy** makes the best local choice (e.g. always keep the interval that ends earliest) and never looks back — when that provably yields the global best.

---

**Prerequisite knowledge:**
- Sorting by end time.
- The activity-selection greedy.

## 📝 Problem

Return the minimum number of intervals to remove so the rest don't overlap.

**Example**
```
[[1,2],[2,3],[3,4],[1,3]] -> 1   (remove [1,3])
```

> One approach: greedy by end time, `O(n log n)`.

### Approach — Greedy by End Time

**Idea:** Sort by end. Track the end of the last kept interval. If the next starts before that end, it overlaps → remove it; otherwise keep it and update the end.

**Time:** `O(n log n)`. **Space:** `O(1)`.

````python
def erase_overlap_intervals(intervals):
    intervals.sort(key=lambda x: x[1])     # sort by EARLIEST finish time
    removed = 0
    prev_end = float("-inf")               # end time of the last interval we kept
    for start, end in intervals:
        if start >= prev_end:              # doesn't overlap the last kept one
            prev_end = end                 # keep it; update the boundary
        else:
            removed += 1                   # overlaps -> we must remove this one
    return removed
````

````python
# Correctness check
tests = [
    ([[1,2],[2,3],[3,4],[1,3]],1),
    ([[1,2],[1,2],[1,2]],2),
    ([[1,2],[2,3]],0),
]
for iv, exp in tests:
    got = erase_overlap_intervals([x[:] for x in iv])
    print(f"{iv} -> remove {got}")
    assert got == exp, "mismatch!"
print("\nAll tests passed")
````

## 🧩 Patterns Learned

- **Greedy by earliest finish:** keeping the interval that ends soonest maximizes how many you keep.
- **Signal:** "fewest removals to make non-overlapping", "max non-overlapping intervals".
- **Related problems:** Meeting Rooms, Minimum Arrows to Burst Balloons, Activity Selection.
- **Common pitfalls:** (1) sorting by start instead of end; (2) `>=` vs `>` at touching endpoints.
