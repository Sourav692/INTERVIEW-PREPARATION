# 70. Climbing Stairs
**Difficulty:** 🟢 Easy · **Topic:** Dynamic Programming · **LeetCode:** https://leetcode.com/problems/climbing-stairs/

## 💡 Concepts

**Core concept(s):** The simplest DP — the number of ways to reach step `n` is the sum of the ways to reach the two steps below it.

**Why it applies here:** To land on step `n` your last move came from step `n-1` (a 1-step) or `n-2` (a 2-step). So `ways(n) = ways(n-1) + ways(n-2)` — the Fibonacci recurrence. Naively recursing recomputes the same steps exponentially; DP computes each once.

**Key intuition:** The ways to reach a step = ways to the step below + ways to two steps below.

---

### 📚 What is Dynamic Programming (DP)?
**DP** solves a big problem by solving smaller **overlapping** subproblems once and reusing the answers. Two styles: **memoization** (recursion that caches results) and **tabulation** (fill a table from the smallest cases up).
- **Why it's fast:** it turns exponential re-computation into a single sweep over the subproblems.
- **In Python:** a `dict`/list cache, or a `dp` list/2-D table.

### 📚 Subproblems & Recurrence
The heart of DP is a **recurrence**: the answer for a state written in terms of smaller states (e.g. `dp[i] = dp[i-1] + dp[i-2]`). Find the recurrence and the base cases, and the code writes itself.

---

**Prerequisite knowledge:**
- Recurrence + base cases.
- Keeping only the last two results.

## 📝 Problem

You climb `n` stairs, 1 or 2 steps at a time. How many distinct ways to reach the top?

**Example**
```
n = 3 -> 3   (1+1+1, 1+2, 2+1)
```

> Three approaches: exponential brute force, O(n) table, and O(1)-space rolling.

### Approach 1 — Brute Recursion (worst)

**Idea:** Directly recurse `ways(n) = ways(n-1) + ways(n-2)`. Recomputes the same values over and over.

**Time:** `O(2^n)`. **Space:** `O(n)`.

````python
def climb_brute(n: int) -> int:
    if n <= 2:
        return n                           # 1 way for 1 stair, 2 ways for 2 stairs
    # Ways to reach n = ways from (n-1) taking a 1-step + ways from (n-2) taking a 2-step.
    return climb_brute(n - 1) + climb_brute(n - 2)   # recomputes the same values (slow)
````

### Approach 2 — DP Table (better)

**Idea:** Fill `dp[i]` from the bottom up so each value is computed once.

**Time:** `O(n)`. **Space:** `O(n)`.

````python
def climb_dp(n: int) -> int:
    if n <= 2:
        return n
    dp = [0] * (n + 1)                     # dp[i] = number of ways to reach stair i
    dp[1], dp[2] = 1, 2                     # base cases
    for i in range(3, n + 1):
        dp[i] = dp[i - 1] + dp[i - 2]      # each stair sums the two below it
    return dp[n]
````

### Approach 3 — Rolling Two Variables (optimal)

**Idea:** Only the last two values matter, so keep just two numbers.

**Time:** `O(n)`. **Space:** `O(1)`.

````python
def climb_opt(n: int) -> int:
    if n <= 2:
        return n
    a, b = 1, 2                            # only the last two results are ever needed
    for _ in range(3, n + 1):
        a, b = b, a + b                    # slide the window forward (like Fibonacci)
    return b
````

````python
# Correctness check
tests = [(1,1),(2,2),(3,3),(4,5),(5,8),(10,89)]
for n, exp in tests:
    a, b, c = climb_brute(n), climb_dp(n), climb_opt(n)
    print(f"n={n} -> brute={a}, dp={b}, opt={c} | expected={exp}")
    assert a == b == c == exp, "mismatch!"
print("\nAll tests passed")
````

## ⏱️ Empirically Checking the Complexities

Big-O can't be read off a function directly, but it can be **measured**. We time each approach on inputs of growing size `n` and read the **doubling ratio** — how much runtime grows when `n` doubles.

| Theoretical | Ratio when `n` → `2n` |
|-------------|-----------------------|
| `O(n)`        | ≈ **2×** |
| `O(n log n)`  | ≈ **2×** (slightly more) |
| `O(n²)`       | ≈ **4×** |

Inputs are shaped to force the worst case. (Exponential brute-force versions are shown in the code but omitted from timing where they would blow up — noted per notebook.)

*(The exponential brute force is omitted here; we compare the two efficient approaches.)*

````python
import os, sys
_root = os.getcwd()
for _ in range(5):
    if os.path.exists(os.path.join(_root, "bench_utils.py")):
        break
    _root = os.path.dirname(_root)
if _root not in sys.path:
    sys.path.insert(0, _root)
from bench_utils import benchmark

def make_worst_case(n):
    return (n,)
solutions = {
    "dp table   O(n) time O(n) space": climb_dp,
    "rolling    O(n) time O(1) space": climb_opt,
}
sizes = [20000, 40000, 80000, 160000]

benchmark(solutions, make_worst_case, sizes, plot=True)
````

## 🧩 Patterns Learned

- **Fibonacci-style recurrence:** "current = combine the previous few" is the most common 1-D DP shape.
- **Shrink the memory:** if `dp[i]` only needs the last couple of values, drop the array for a few variables.
- **Signal:** "count the ways to reach / build something in steps".
- **Related problems:** House Robber, Decode Ways, Min Cost Climbing Stairs, Fibonacci.
- **Common pitfalls:** (1) recursion without caching (exponential); (2) wrong base cases.
