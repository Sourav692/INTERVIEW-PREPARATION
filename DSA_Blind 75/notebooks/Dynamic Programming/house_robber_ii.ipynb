## 💡 Concepts

**Core concept(s):** Reduce a **circular** version to the plain House Robber — run it twice.

**Why it applies here:** The houses form a **circle**, so the first and last are adjacent and can't both be robbed. Split into two straight-line cases — rob houses `0..n-2` (exclude the last) or `1..n-1` (exclude the first) — solve each with the linear robber, and take the better.

**Key intuition:** First and last touch, so just try 'without the last' and 'without the first' and keep the bigger.

---

### 📚 What is Dynamic Programming (DP)?
**DP** solves a big problem by solving smaller **overlapping** subproblems once and reusing the answers. Two styles: **memoization** (recursion that caches results) and **tabulation** (fill a table from the smallest cases up).
- **Why it's fast:** it turns exponential re-computation into a single sweep over the subproblems.
- **In Python:** a `dict`/list cache, or a `dp` list/2-D table.

### 📚 Subproblems & Recurrence
The heart of DP is a **recurrence**: the answer for a state written in terms of smaller states (e.g. `dp[i] = dp[i-1] + dp[i-2]`). Find the recurrence and the base cases, and the code writes itself.

**Prerequisite knowledge:**
- The linear House Robber solution.
- Reducing a new problem to a solved one.

## 📝 Problem

Same as House Robber, but the houses are arranged in a **circle** (first and last are neighbors).

**Example**
```
[2,3,2] -> 3   (can't take both 2s at the ends)
[1,2,3,1] -> 4
```

> One approach: run the linear robber on two ranges and take the max.

### Approach — Linear Robber Twice

**Idea:** The answer is the better of robbing `nums[:-1]` (drop the last house) and `nums[1:]` (drop the first). Handle the single-house case directly.

**Time:** `O(n)`. **Space:** `O(1)`.

````python
def rob_line(arr):                          # plain House Robber on a straight line
    prev, cur = 0, 0
    for x in arr:
        prev, cur = cur, max(cur, prev + x)  # skip, or rob (x + best two-back)
    return cur

def rob2(nums):
    if len(nums) == 1:
        return nums[0]                     # a single house: just rob it
    # Houses form a circle, so the first and last can't both be robbed.
    # Try "without the last house" and "without the first house"; take the better.
    return max(rob_line(nums[:-1]), rob_line(nums[1:]))
````

````python
# Correctness check
tests = [([2,3,2],3), ([1,2,3,1],4), ([1,2,3],3), ([5],5), ([1,2],2)]
for nums, exp in tests:
    got = rob2(nums)
    print(f"{nums} -> {got} | expected={exp}")
    assert got == exp, "mismatch!"
print("\nAll tests passed")
````

## ⏱️ Empirically Checking the Complexities

Big-O can't be read off a function directly, but it can be **measured**. We time each approach on inputs of growing size `n` and read the **doubling ratio** — how much runtime grows when `n` doubles.

| Theoretical | Ratio when `n` → `2n` |
|-------------|-----------------------|
| `O(n)`        | ≈ **2×** |
| `O(n log n)`  | ≈ **2×** (slightly more) |
| `O(n²)`       | ≈ **4×** |

Inputs are shaped to force the worst case. (Exponential brute-force versions are shown in the code but omitted from timing where they would blow up — noted per notebook.)

````python
import os, sys
_root = os.getcwd()
for _ in range(5):
    if os.path.exists(os.path.join(_root, "bench_utils.py")):
        break
    _root = os.path.dirname(_root)
if _root not in sys.path:
    sys.path.insert(0, _root)
from bench_utils import benchmark

def make_worst_case(n):
    return ([(i * 37) % 100 for i in range(n)],)
solutions = {
    "linear robber x2 O(n)": rob2,
}
sizes = [20000, 40000, 80000, 160000]

benchmark(solutions, make_worst_case, sizes, plot=True)
````

## 🧩 Patterns Learned

- **Reduce to a solved problem:** a circular constraint becomes two linear cases.
- **Case-split on the tricky element:** try with and without the conflicting house.
- **Signal:** "same as X but arranged in a circle / with a wrap-around".
- **Related problems:** House Robber, Maximum Sum Circular Subarray.
- **Common pitfalls:** (1) forgetting the single-house edge case; (2) double-counting the wrap-around pair.
