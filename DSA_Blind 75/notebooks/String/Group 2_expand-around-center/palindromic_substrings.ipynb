# 647. Palindromic Substrings
**Difficulty:** 🟡 Medium · **Topic:** String · **LeetCode:** https://leetcode.com/problems/palindromic-substrings/

## 💡 Concepts

**Core concept(s):** **Expand around center** — count palindromes as you stretch out from each middle.

**Why it applies here:** Instead of finding the single longest palindrome, we *count them all*. Each center (a character, or a gap) gives one new palindrome for every successful outward step, so counting is a natural by-product of expanding.

**Key intuition:** Every time an expansion from a center still matches, that's one more palindrome.

---

### 📚 What is "Expand Around Center"?
Every palindrome has a middle. **Expand around center** places a middle (a single character, or the gap between two) and walks outward while the two sides match.
- **Complexity:** each of the ~2n centers can expand up to n steps → **O(n²)** time, **O(1)** space.

---

**Prerequisite knowledge:**
- Expanding from centers and counting successful steps.

## 📝 Problem

Count how many substrings of `s` are palindromes (each position counts).

**Example**
```
"abc" -> 3    ("a","b","c")
"aaa" -> 6    ("a","a","a","aa","aa","aaa")
```

> Two approaches: brute `O(n^3)` and expand-around-center `O(n^2)`.

### Approach 1 — Check Every Substring (worst)

**Idea:** Test every substring; count the palindromes.

**Time complexity:** `O(n^3)`.

**Space complexity:** `O(1)` extra.

````python
def count_pal_brute(s: str) -> int:
    n = len(s)
    total = 0
    for i in range(n):                     # substring start
        for j in range(i, n):              # substring end
            sub = s[i:j+1]
            if sub == sub[::-1]:           # reads the same reversed -> a palindrome
                total += 1
    return total
````

### Approach 2 — Expand Around Center (optimal)

**Idea:** From each center, every successful outward step is one more palindrome. Sum over all odd and even centers.

**Time complexity:** `O(n^2)`.

**Space complexity:** `O(1)`.

````python
def count_pal_expand(s: str) -> int:
    n = len(s)
    def grow(l, r):                        # count palindromes centered at (l, r)
        cnt = 0
        while l >= 0 and r < n and s[l] == s[r]:
            cnt += 1                        # each successful expansion is one more palindrome
            l -= 1; r += 1                  # widen outward
        return cnt
    total = 0
    for i in range(n):
        total += grow(i, i)                 # odd-length palindromes centered on i
        total += grow(i, i + 1)             # even-length palindromes centered between i, i+1
    return total
````

````python
# Correctness check
tests = [("abc",3), ("aaa",6), ("a",1), ("aba",4), ("",0)]
for s, exp in tests:
    a, b = count_pal_brute(s), count_pal_expand(s)
    print(f"{s!r:>6} -> brute={a}, expand={b} | expected={exp}")
    assert a == b == exp, "mismatch!"
print("\nAll tests passed")
````

## ⏱️ Empirically Checking the Complexities

Big-O can't be read off a function directly, but it can be **measured**. We time each approach on inputs of growing `n` and read the **doubling ratio** — how much runtime grows when `n` doubles.

| Theoretical | Ratio when `n` → `2n` |
|-------------|-----------------------|
| `O(n)`        | ≈ **2×** |
| `O(n log n)`  | ≈ **2×** (slightly more) |
| `O(n²)`       | ≈ **4×** |
| `O(n³)`       | ≈ **8×** |

Inputs are built to force the **worst case** (no early exit). Sub-millisecond rows are noisy — look at the trend.

````python
import os, sys
_root = os.getcwd()
for _ in range(5):
    if os.path.exists(os.path.join(_root, "bench_utils.py")):
        break
    _root = os.path.dirname(_root)
if _root not in sys.path:
    sys.path.insert(0, _root)
from bench_utils import benchmark   # shared: prints ratio table + optional log-log plot

def make_worst_case(n):
    s = "a" * n                             # every substring is a palindrome -> worst case
    return (s,)

solutions = {
    "brute  O(n^3)": count_pal_brute,
    "expand O(n^2)": count_pal_expand,
}
sizes = [40, 80, 160, 320]                  # small: brute is cubic

benchmark(solutions, make_worst_case, sizes, plot=True)
````

## 🧩 Patterns Learned

- **Counting via expansion:** the same "expand around center" engine that finds the longest palindrome also counts them all.
- **Signal:** "how many palindromic substrings", "count symmetric pieces".
- **Related problems:** Longest Palindromic Substring, Palindromic Subsequences.
- **Common pitfalls:** (1) forgetting even-length centers; (2) double-counting single characters.
