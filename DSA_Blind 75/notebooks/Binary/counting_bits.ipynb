# 338. Counting Bits
**Difficulty:** 🟢 Easy · **Topic:** Binary · **LeetCode:** https://leetcode.com/problems/counting-bits/

## 💡 Concepts

**Core concept(s):** DP on bits — the number of 1s in `i` equals the 1s in `i >> 1` plus its last bit.

**Why it applies here:** Dropping the last bit of `i` (i.e. `i // 2`) gives a smaller number we've already counted; add back 1 if `i` was odd. That reuses earlier answers, so the whole array fills in one pass.

**Key intuition:** bits(i) = bits(i without its last bit) + (last bit).

---

### 📚 What is Binary / Bits?
Computers store numbers in **binary** — strings of 0/1 **bits**. Bit operations act on those directly: `&` (and), `|` (or), `^` (xor), `~` (not), `<<`/`>>` (shift). They are extremely fast.

### 📚 Masks & Shifting
A **mask** like `0xFFFFFFFF` keeps only certain bits; shifting (`<<`, `>>`) moves bits to inspect or build a number one bit at a time.

---

**Prerequisite knowledge:**
- The `i >> 1` relationship.
- Simple 1-D DP.

## 📝 Problem

For `0..n`, return an array where entry `i` is the number of 1 bits in `i`.

**Example**
```
n=5 -> [0,1,1,2,1,2]
```

> Two approaches: count each number `O(n log n)` and DP `O(n)`.

### Approach 1 — Count Each (worst)

**Idea:** Count the bits of every number independently.

**Time:** `O(n log n)`. **Space:** `O(n)`.

````python
def counting_brute(n: int):
    # For each number, count its 1 bits independently.
    return [bin(i).count("1") for i in range(n + 1)]
````

### Approach 2 — DP with `i >> 1` (optimal)

**Idea:** `dp[i] = dp[i >> 1] + (i & 1)` — reuse the already-computed smaller number.

**Time:** `O(n)`. **Space:** `O(n)`.

````python
def counting_dp(n: int):
    dp = [0] * (n + 1)
    for i in range(1, n + 1):
        # i without its last bit is i >> 1 (already counted); add back the last bit.
        dp[i] = dp[i >> 1] + (i & 1)
    return dp
````

````python
# Correctness check
for n, exp in [(2,[0,1,1]),(5,[0,1,1,2,1,2]),(0,[0])]:
    a, b = counting_brute(n), counting_dp(n)
    print(f"n={n} -> {b}")
    assert a == b == exp, "mismatch!"
print("\nAll tests passed")
````

## 🧩 Patterns Learned

- **DP over a bit relationship:** reuse `i >> 1` to build counts in one pass.
- **Signal:** "bit counts for a whole range", "popcount 0..n".
- **Related problems:** Number of 1 Bits, Divide numbers by 2 patterns.
- **Common pitfalls:** (1) recomputing each number from scratch; (2) off-by-one on the array length (`n+1`).
