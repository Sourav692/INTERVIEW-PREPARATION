# 268. Missing Number
**Difficulty:** 🟢 Easy · **Topic:** Binary · **LeetCode:** https://leetcode.com/problems/missing-number/

## 💡 Concepts

**Core concept(s):** Find the one missing value from `0..n` via the **sum formula** or by **XOR**.

**Why it applies here:** The full set `0..n` has a known sum; subtract the actual sum to get the gap. Or XOR all indices and values together — every present number cancels, leaving the missing one.

**Key intuition:** Compare the array against the complete set 0..n — the difference (or the XOR leftover) is the gap.

---

### 📚 What is Binary / Bits?
Computers store numbers in **binary** — strings of 0/1 **bits**. Bit operations act on those directly: `&` (and), `|` (or), `^` (xor), `~` (not), `<<`/`>>` (shift). They are extremely fast.

### 📚 The XOR Trick
`^` (XOR) gives 1 only when bits differ. Two key facts: `x ^ x = 0` and `x ^ 0 = x`. So XORing a value in twice cancels it — great for finding the one unmatched item.

---

**Prerequisite knowledge:**
- Arithmetic sum, or the XOR-cancels-pairs trick.

## 📝 Problem

An array holds `n` distinct numbers from `0..n`; return the one that's missing.

**Example**
```
[3,0,1] -> 2
```

> Two approaches, both `O(n)`: sum formula and XOR.

### Approach 1 — Sum Formula

**Idea:** Expected sum of `0..n` minus the actual sum is the missing number.

**Time:** `O(n)`. **Space:** `O(1)`.

````python
def missing_sum(nums):
    n = len(nums)
    # Sum of 0..n minus the sum we actually have = the missing number.
    return n * (n + 1) // 2 - sum(nums)
````

### Approach 2 — XOR (optimal)

**Idea:** XOR every index `0..n` and every value. Matching pairs cancel; the missing number survives.

**Time:** `O(n)`. **Space:** `O(1)`.

````python
def missing_xor(nums):
    x = len(nums)                          # start with n (the largest possible index)
    for i, v in enumerate(nums):
        x ^= i ^ v                         # XOR each index with each value; equal ones cancel
    return x                               # the unmatched leftover is the missing number
````

````python
# Correctness check
tests = [([3,0,1],2),([0,1],2),([9,6,4,2,3,5,7,0,1],8),([0],1)]
for nums, exp in tests:
    a, b = missing_sum(nums), missing_xor(nums)
    print(f"{nums} -> sum={a}, xor={b} | expected={exp}")
    assert a == b == exp, "mismatch!"
print("\nAll tests passed")
````

## 🧩 Patterns Learned

- **XOR cancels pairs:** pairing indices with values isolates the odd one out with O(1) memory and no overflow.
- **Signal:** "one number missing / appears once", "find the odd one".
- **Related problems:** Single Number, Find the Difference, Find All Numbers Disappeared.
- **Common pitfalls:** (1) integer overflow with sums in fixed-width languages (XOR avoids it).
