# 191. Number of 1 Bits
**Difficulty:** 🟢 Easy · **Topic:** Binary · **LeetCode:** https://leetcode.com/problems/number-of-1-bits/

## 💡 Concepts

**Core concept(s):** Count set bits by **shifting** through them, or clear them one at a time with `n & (n-1)`.

**Why it applies here:** Each `1` bit contributes to the count. Either test the last bit and shift right (32 steps), or use the trick that `n & (n-1)` removes the lowest set bit — so the loop runs once per 1 bit.

**Key intuition:** Either check each bit position, or keep knocking out the lowest 1 until none remain.

---

### 📚 What is Binary / Bits?
Computers store numbers in **binary** — strings of 0/1 **bits**. Bit operations act on those directly: `&` (and), `|` (or), `^` (xor), `~` (not), `<<`/`>>` (shift). They are extremely fast.

### 📚 Masks & Shifting
A **mask** like `0xFFFFFFFF` keeps only certain bits; shifting (`<<`, `>>`) moves bits to inspect or build a number one bit at a time.

**Prerequisite knowledge:**
- Bit shifting and `&`.
- The `n & (n-1)` trick.

## 📝 Problem

Count the number of `1` bits in an integer.

**Example**
```
11 (1011) -> 3
```

> Two approaches: shift-through-all-bits and Brian Kernighan's clear-lowest-bit.

### Approach 1 — Shift Every Bit

**Idea:** Test the last bit, shift right, repeat.

**Time:** `O(bits)`. **Space:** `O(1)`.

````python
def hamming_loop(n: int) -> int:
    count = 0
    while n:
        count += n & 1                     # add 1 if the last bit is set
        n >>= 1                            # drop the last bit and continue
    return count
````

### Approach 2 — Clear Lowest Bit (optimal)

**Idea:** `n & (n-1)` removes the lowest set bit; count how many times you can do it.

**Time:** `O(number of 1 bits)`. **Space:** `O(1)`.

````python
def hamming_kernighan(n: int) -> int:
    count = 0
    while n:
        n &= n - 1                         # clears the lowest set bit each time
        count += 1                         # so the loop runs once per 1 bit
    return count
````

````python
# Correctness check
tests = [(11,3),(128,1),(0,0),(255,8),(2147483647,31)]
for n, exp in tests:
    a, b = hamming_loop(n), hamming_kernighan(n)
    print(f"{n} -> loop={a}, kernighan={b} | expected={exp}")
    assert a == b == exp, "mismatch!"
print("\nAll tests passed")
````

## ⏱️ Empirically Checking the Complexities

We time each approach on inputs of growing size `n` and read the **doubling ratio**.

| Theoretical | Ratio `n`→`2n` |
|---|---|
| `O(n)`       | ≈ **2×** |
| `O(n log n)` | ≈ **2×** (slightly more) |
| `O(n²)`     | ≈ **4×** |

````python
import os, sys
_root = os.getcwd()
for _ in range(6):
    if os.path.exists(os.path.join(_root, "bench_utils.py")): break
    _root = os.path.dirname(_root)
if _root not in sys.path: sys.path.insert(0, _root)
from bench_utils import benchmark

def make_worst_case(n):
    return ((1 << n) - 1,)   # all n bits set -> both loops run n times
solutions = {
    "shift     O(bits)": hamming_loop,
    "kernighan O(1-bits)": hamming_kernighan,
}
sizes = [2000, 4000, 8000, 16000]

benchmark(solutions, make_worst_case, sizes, plot=True)
````
