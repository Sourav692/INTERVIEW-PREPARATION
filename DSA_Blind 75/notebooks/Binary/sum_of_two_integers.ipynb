## 💡 Concepts

**Core concept(s):** Add without `+` using **XOR** (add-without-carry) and **AND+shift** (the carry).

**Why it applies here:** Adding two bits: XOR gives the sum bit ignoring carry; AND then shift-left gives where a carry lands. Repeat until there's no carry left — that's binary addition by hand.

**Key intuition:** XOR adds the bits, AND«shift finds the carry; loop until no carry remains.

---

### 📚 What is Binary / Bits?
Computers store numbers in **binary** — strings of 0/1 **bits**. Bit operations act on those directly: `&` (and), `|` (or), `^` (xor), `~` (not), `<<`/`>>` (shift). They are extremely fast.

### 📚 The XOR Trick
`^` (XOR) gives 1 only when bits differ. Two key facts: `x ^ x = 0` and `x ^ 0 = x`. So XORing a value in twice cancels it — great for finding the one unmatched item.

### 📚 Masks & Shifting
A **mask** like `0xFFFFFFFF` keeps only certain bits; shifting (`<<`, `>>`) moves bits to inspect or build a number one bit at a time.

---

**Prerequisite knowledge:**
- XOR and AND on bits.
- A 32-bit mask to emulate fixed-width integers in Python.

## 📝 Problem

Return `a + b` **without** using `+` or `-`.

**Example**
```
1 + 2 = 3 ;  -2 + 3 = 1
```

> One approach: bit manipulation (with a 32-bit mask for negatives).

### Approach — XOR + Carry

**Idea:** `a ^ b` sums the bits; `(a & b) << 1` is the carry. Loop until carry is 0. The mask keeps everything in 32-bit range; negatives are recovered at the end.

**Time:** `O(1)` (≤ 32 iterations). **Space:** `O(1)`.

````python
def get_sum(a: int, b: int) -> int:
    mask = 0xFFFFFFFF                      # keep numbers to 32 bits (Python ints are unbounded)
    while b != 0:                          # repeat until there is no carry left
        carry = (a & b) << 1               # a carry happens where BOTH bits are 1, shifted left
        a = (a ^ b) & mask                 # XOR adds the bits WITHOUT the carry
        b = carry & mask                   # apply the carry on the next round
    return a if a <= 0x7FFFFFFF else ~(a ^ mask)   # convert back to a signed (negative) number
````

````python
# Correctness check
tests = [(1,2,3),(-2,3,1),(0,0,0),(-1,1,0),(20,-8,12)]
for a, b, exp in tests:
    got = get_sum(a, b)
    print(f"{a} + {b} = {got} | expected={exp}")
    assert got == exp, "mismatch!"
print("\nAll tests passed")
````

*(We time a non-negative adder as the number of bits grows, to show the per-bit cost.)*

````python
import os, sys
_root = os.getcwd()
for _ in range(6):
    if os.path.exists(os.path.join(_root, "bench_utils.py")): break
    _root = os.path.dirname(_root)
if _root not in sys.path: sys.path.insert(0, _root)
from bench_utils import benchmark

def add_big(a, b):
    while b:
        carry = (a & b) << 1
        a ^= b
        b = carry
    return a

def make_worst_case(n):
    return ((1 << n) - 1, 1)   # forces a carry to ripple through all n bits
solutions = {
    "bit add O(bits)": add_big,
}
sizes = [2000, 4000, 8000, 16000]

benchmark(solutions, make_worst_case, sizes, plot=True)
````
