# 190. Reverse Bits
**Difficulty:** 🟢 Easy · **Topic:** Binary · **LeetCode:** https://leetcode.com/problems/reverse-bits/

## 💡 Concepts

**Core concept(s):** Build the reversed number one bit at a time by shifting.

**Why it applies here:** Reversing bits means the last bit becomes the first, and so on. Pull the lowest bit off the input and push it onto a result that you shift left each step — after 32 steps the result is the mirror image.

**Key intuition:** Peel the input's last bit and stack it onto the front of the result.

---

### 📚 What is Binary / Bits?
Computers store numbers in **binary** — strings of 0/1 **bits**. Bit operations act on those directly: `&` (and), `|` (or), `^` (xor), `~` (not), `<<`/`>>` (shift). They are extremely fast.

### 📚 Masks & Shifting
A **mask** like `0xFFFFFFFF` keeps only certain bits; shifting (`<<`, `>>`) moves bits to inspect or build a number one bit at a time.

**Prerequisite knowledge:**
- Shifting bits in and out.

## 📝 Problem

Reverse the bits of a 32-bit unsigned integer.

**Example**
```
0000...0101 (5) -> 1010...0000
```

> One approach: shift bits out of the input and into the result.

### Approach — Shift In / Shift Out

**Idea:** 32 times: shift `result` left, OR in the input's last bit, shift the input right.

**Time:** `O(32) = O(1)`. **Space:** `O(1)`.

````python
def reverse_bits(n: int, width: int = 32) -> int:
    res = 0
    for _ in range(width):                 # process each of the `width` bits
        res = (res << 1) | (n & 1)         # shift result left, drop in the input's last bit
        n >>= 1                            # move to the input's next bit
    return res
````

````python
# Correctness check (small widths are easy to verify by hand)
assert reverse_bits(0b101, 3) == 0b101          # palindrome
assert reverse_bits(0b110, 3) == 0b011
assert reverse_bits(43261596, 32) == 964176192  # LeetCode example
print("All tests passed")
````

*(We reverse a `width`-bit number and grow the width to show the per-bit cost.)*
