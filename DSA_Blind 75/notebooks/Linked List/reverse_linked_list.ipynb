# 206. Reverse Linked List
**Difficulty:** 🟢 Easy · **Topic:** Linked List · **LeetCode:** https://leetcode.com/problems/reverse-linked-list/

## 💡 Concepts

**Core concept(s):** Walk the list flipping each `next` pointer to point backward.

**Why it applies here:** Reversing means every arrow should point the other way. Carry a `prev` pointer; at each node, remember the next node, point the current node back at `prev`, then advance.

**Key intuition:** Flip one arrow at a time, dragging a 'prev' pointer along.

---

### 📚 What is a Linked List?
A **linked list** is a chain of nodes; each node holds a value and a pointer to the **next** node. Unlike an array there is no index — you can only walk forward from the head.
- **In Python:** a small `ListNode` class with `.val` and `.next`.

### 📚 Pointers & the Dummy Node
Linked-list code moves **pointers** (references to nodes). A **dummy** node placed before the head removes annoying "is this the first node?" special cases — you build off `dummy.next` and return it at the end.

### 📚 Recursion on a List
A list problem often splits into "handle this node, then recurse on the rest". Clean to write, but uses stack depth equal to the list length.

---

**Prerequisite knowledge:**
- Pointer juggling.
- Optional recursion.

## 📝 Problem

Reverse a singly linked list; return the new head.

**Example**
```
1->2->3->4->5  ->  5->4->3->2->1
```

> Two approaches, both `O(n)`: iterative (O(1) space) and recursive (O(n) stack).

````python
from typing import Optional, List

class ListNode:
    """A node in a singly linked list: a value plus a link to the next node."""
    def __init__(self, val=0, next=None):
        self.val = val                     # the value stored at this node
        self.next = next                   # link to the next node (None at the end)

def build_list(vals):
    """Turn a Python list into a linked list; return its head."""
    dummy = ListNode(); cur = dummy        # dummy node avoids special-casing the first node
    for v in vals:
        cur.next = ListNode(v); cur = cur.next
    return dummy.next

def to_list(head):
    """Turn a linked list back into a Python list (handy for printing / assertions)."""
    out = []
    while head:
        out.append(head.val); head = head.next
    return out
````

### Approach 1 — Iterative (optimal)

**Idea:** Keep `prev`. At each node, save `next`, point `node.next` to `prev`, move both forward.

**Time:** `O(n)`. **Space:** `O(1)`.

````python
def reverse_iter(head: Optional[ListNode]) -> Optional[ListNode]:
    prev = None                            # the reversed part starts empty
    while head:
        nxt = head.next                    # remember the rest before we overwrite the link
        head.next = prev                   # flip this node's arrow to point backward
        prev = head                        # prev moves up to this node
        head = nxt                         # continue with the rest of the list
    return prev                            # prev is the new head
````

### Approach 2 — Recursive

**Idea:** Reverse the rest, then make the next node point back at this one.

**Time:** `O(n)`. **Space:** `O(n)` recursion stack.

````python
def reverse_rec(head: Optional[ListNode]) -> Optional[ListNode]:
    if not head or not head.next:
        return head                        # empty or single node: already reversed
    new_head = reverse_rec(head.next)      # reverse everything after this node first
    head.next.next = head                  # make the next node point back at us
    head.next = None                       # we become the new tail
    return new_head
````

````python
# Correctness check
for fn in (reverse_iter, reverse_rec):
    for vals in ([1,2,3,4,5], [1], []):
        assert to_list(fn(build_list(vals))) == vals[::-1]
    print(fn.__name__, "OK")
print("\nAll tests passed")
````

## ⏱️ Empirically Checking the Complexities

We time each approach on inputs of growing size `n` and read the **doubling ratio**.

| Theoretical | Ratio when `n` → `2n` |
|---|---|
| `O(n)`       | ≈ **2×** |
| `O(n log n)` | ≈ **2×** (slightly more) |
| `O(n²)`     | ≈ **4×** |

*(Recursive version is O(n) too but limited by stack depth; we time the iterative one as the list grows.)*

````python
import os, sys
_root = os.getcwd()
for _ in range(5):
    if os.path.exists(os.path.join(_root, "bench_utils.py")): break
    _root = os.path.dirname(_root)
if _root not in sys.path: sys.path.insert(0, _root)
from bench_utils import benchmark

def make_worst_case(n):
    return (build_list(list(range(n))),)
solutions = {
    "iterative O(n)": reverse_iter,
}
sizes = [20000, 40000, 80000, 160000]

benchmark(solutions, make_worst_case, sizes, plot=True)
````
