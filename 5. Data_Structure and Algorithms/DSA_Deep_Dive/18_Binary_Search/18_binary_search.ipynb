The classic search, `bisect_left`/`bisect_right`, and binary search on the answer (a monotonic predicate).

## 1. Classic binary search -- O(log n)

```python
def binary_search(arr, target):
    lo, hi = 0, len(arr) - 1
    while lo <= hi:
        mid = lo + (hi - lo) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            lo = mid + 1
        else:
            hi = mid - 1
    return -1

arr = [1, 3, 5, 7, 9, 11, 13]
print("search for 11:", binary_search(arr, 11))
print("search for 4 (absent):", binary_search(arr, 4))
assert binary_search(arr, 11) == 5
assert binary_search(arr, 1) == 0            # first element
assert binary_search(arr, 13) == 6           # last element
assert binary_search(arr, 4) == -1           # not present
assert binary_search([], 5) == -1            # empty array
```

## 2. `bisect_left` vs `bisect_right`

```python
import bisect

a = [1, 3, 3, 3, 7]
left = bisect.bisect_left(a, 3)
right = bisect.bisect_right(a, 3)
print("bisect_left(a, 3) ->", left)
print("bisect_right(a, 3) ->", right)
assert left == 1     # before all the 3s
assert right == 4    # after all the 3s

# The "latest entry <= x" pattern -- used for checkpoint-style queries.
def latest_at_or_before(a, x):
    idx = bisect.bisect_right(a, x) - 1
    return a[idx] if idx >= 0 else None

assert latest_at_or_before(a, 3) == 3
assert latest_at_or_before(a, 5) == 3    # no exact 5 -- falls back to the latest <= 5, which is 3
assert latest_at_or_before(a, 0) is None  # nothing is <= 0
```

## 3. Binary search on the answer -- a monotonic predicate, no array needed

```python
def binary_search_answer(lo, hi, predicate):
    """Find the smallest x in [lo, hi] where predicate(x) is True, given False...False True...True."""
    while lo < hi:
        mid = lo + (hi - lo) // 2
        if predicate(mid):
            hi = mid            # mid works -- see if something smaller also works
        else:
            lo = mid + 1         # mid doesn't work -- must go higher
    return lo


# Classic example: "first bad version" -- versions 1..n, some cutoff version and every
# version after it is bad. Find the FIRST bad version with the fewest predicate calls.
FIRST_BAD = 37
def is_bad_version(v):
    return v >= FIRST_BAD

result = binary_search_answer(1, 100, is_bad_version)
print("first bad version:", result)
assert result == FIRST_BAD

# Second example: smallest integer x such that x*x >= 50 (an integer square-root-ish search).
def predicate(x):
    return x * x >= 50

result2 = binary_search_answer(0, 50, predicate)
print("smallest x with x^2 >= 50:", result2)
assert result2 == 8          # 7^2=49 (false), 8^2=64 (true)
```

## ✅ Recap
- Binary search needs **sorted or monotonic** data -- each comparison halves the remaining search space, giving **O(log n)**.
- `bisect_left` finds the leftmost valid insertion point; `bisect_right` the rightmost -- `bisect_right(a, x) - 1` finds the latest entry `<= x`.
- **Binary search on the answer** generalizes the technique to any monotonic yes/no predicate, even with no literal array ("first bad version", integer square root, "minimum capacity that works").
- Watch for off-by-one bugs: `mid +/- 1` (never leave `lo`/`hi` unchanged), and `<=` vs `<` in the loop condition.
