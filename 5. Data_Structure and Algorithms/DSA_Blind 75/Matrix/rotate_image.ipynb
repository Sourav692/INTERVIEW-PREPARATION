# 48. Rotate Image
**Difficulty:** 🟡 Medium · **Topic:** Matrix · **LeetCode:** https://leetcode.com/problems/rotate-image/

## 💡 Concepts

**Core concept(s):** Rotate 90° in place by **transpose then reverse each row**.

**Why it applies here:** Transposing (swap across the diagonal) turns rows into columns; reversing each row then puts them in the rotated position. Two simple in-place passes do it with no extra grid.

**Key intuition:** Flip across the diagonal, then mirror each row left-to-right.

---

### 📚 Working with a Matrix (Grid)
A matrix is a list of rows. Common moves: walk with `(r, c)` coordinates, **transpose** (swap rows/columns), or use the first row/column as scratch space to save memory.

---

**Prerequisite knowledge:**
- Transpose (swap `[i][j]` with `[j][i]`).
- Reversing a row.

## 📝 Problem

Rotate an `n × n` matrix 90° clockwise, **in place**.

**Example**
```
[[1,2,3],[4,5,6],[7,8,9]] -> [[7,4,1],[8,5,2],[9,6,3]]
```

> Two approaches: build a new matrix `O(n²)` space, and transpose+reverse `O(1)` space.

### Approach 1 — New Matrix (worst on memory)

**Idea:** Column `c` from the bottom up becomes row `c` of the result.

**Time:** `O(n²)`. **Space:** `O(n²)`.

````python
def rotate_new(matrix):
    n = len(matrix)
    # New cell (row=c) comes from the old column c read bottom-to-top.
    rotated = [[matrix[n - 1 - r][c] for r in range(n)] for c in range(n)]
    matrix[:] = rotated                    # copy the result back into the original object
    return matrix
````

### Approach 2 — Transpose + Reverse (optimal)

**Idea:** Swap across the diagonal (transpose), then reverse each row.

**Time:** `O(n²)`. **Space:** `O(1)`.

````python
def rotate_inplace(matrix):
    n = len(matrix)
    for i in range(n):                     # step 1: transpose (mirror across the diagonal)
        for j in range(i + 1, n):          # j from i+1 so each pair is swapped once
            matrix[i][j], matrix[j][i] = matrix[j][i], matrix[i][j]
    for row in matrix:                     # step 2: reverse each row
        row.reverse()                      # transpose + reverse = 90 degree clockwise turn
    return matrix
````

````python
# Correctness check
def clone(g): return [row[:] for row in g]
tests = [
    ([[1,2,3],[4,5,6],[7,8,9]],[[7,4,1],[8,5,2],[9,6,3]]),
    ([[1,2],[3,4]],[[3,1],[4,2]]),
]
for g, exp in tests:
    assert rotate_new(clone(g)) == exp
    assert rotate_inplace(clone(g)) == exp
    print("OK", exp)
print("\nAll tests passed")
````

## ⏱️ Empirically Checking the Complexities

We time each approach on inputs of growing size `n` and read the **doubling ratio**.

| Theoretical | Ratio `n`→`2n` |
|---|---|
| `O(n)`       | ≈ **2×** |
| `O(n log n)` | ≈ **2×** (slightly more) |
| `O(n²)`     | ≈ **4×** |

````python
import os, sys
_root = os.getcwd()
for _ in range(6):
    if os.path.exists(os.path.join(_root, "bench_utils.py")): break
    _root = os.path.dirname(_root)
if _root not in sys.path: sys.path.insert(0, _root)
from bench_utils import benchmark

def make_worst_case(n):
    grid = [[r*n + c for c in range(n)] for r in range(n)]
    return (grid,)
solutions = {
    "new matrix O(n^2) space": rotate_new,
    "in-place   O(1) space  ": rotate_inplace,
}
sizes = [100, 200, 400, 800]

benchmark(solutions, make_worst_case, sizes, plot=True)
````

## 🧩 Patterns Learned

- **Transpose + reverse = rotate:** two elementary passes replace an extra grid.
- **Signal:** "rotate a matrix 90° in place".
- **Related problems:** Spiral Matrix, Set Matrix Zeroes, Flip/Invert Image.
- **Common pitfalls:** (1) transposing the whole matrix twice (loop `j` from `i+1`); (2) reversing columns instead of rows for clockwise.
