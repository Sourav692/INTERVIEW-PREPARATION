# 1143. Longest Common Subsequence
**Difficulty:** 🟡 Medium · **Topic:** Dynamic Programming · **LeetCode:** https://leetcode.com/problems/longest-common-subsequence/

## 💡 Concepts

**Core concept(s):** 2-D DP over the two strings' positions.

**Why it applies here:** Compare the two strings position by position. If the current letters match, the common subsequence grows by 1 and we move both forward; if not, we take the better of skipping one letter from either string. Every `(i, j)` pair is a subproblem solved once in a grid.

**Key intuition:** If the two current letters match, add 1 and advance both; otherwise skip a letter from whichever string does better.

---

### 📚 What is Dynamic Programming (DP)?
**DP** solves a big problem by solving smaller **overlapping** subproblems once and reusing the answers. Two styles: **memoization** (recursion that caches results) and **tabulation** (fill a table from the smallest cases up).
- **Why it's fast:** it turns exponential re-computation into a single sweep over the subproblems.
- **In Python:** a `dict`/list cache, or a `dp` list/2-D table.

### 📚 Subproblems & Recurrence
The heart of DP is a **recurrence**: the answer for a state written in terms of smaller states (e.g. `dp[i] = dp[i-1] + dp[i-2]`). Find the recurrence and the base cases, and the code writes itself.

---

**Prerequisite knowledge:**
- Filling a 2-D table.
- The match / skip recurrence.

## 📝 Problem

Return the length of the longest subsequence present in **both** strings.

**Example**
```
"abcde", "ace" -> 3   ("ace")
"abc", "def"   -> 0
```

> Two approaches: exponential brute force and `O(m×n)` DP.

### Approach 1 — Brute Recursion (worst)

**Idea:** At each pair of positions, either the letters match (take it, advance both) or try skipping from each side.

**Time:** `O(2^(m+n))`. **Space:** `O(m+n)`.

````python
def lcs_brute(a, b):
    def dfs(i, j):                         # LCS length of a[i:] and b[j:]
        if i == len(a) or j == len(b):
            return 0                       # one string ran out
        if a[i] == b[j]:
            return 1 + dfs(i + 1, j + 1)   # letters match -> take both, move on
        return max(dfs(i + 1, j), dfs(i, j + 1))   # else skip a letter from one side
    return dfs(0, 0)
````

### Approach 2 — 2-D DP (optimal)

**Idea:** `dp[i][j]` = LCS of `a[i:]` and `b[j:]`. Fill from the bottom-right corner using the match/skip rule.

**Time:** `O(m×n)`. **Space:** `O(m×n)`.

````python
def lcs_dp(a, b):
    m, n = len(a), len(b)
    dp = [[0] * (n + 1) for _ in range(m + 1)]   # dp[i][j] = LCS of a[i:] and b[j:]
    for i in range(m - 1, -1, -1):         # fill from the bottom-right corner
        for j in range(n - 1, -1, -1):
            if a[i] == b[j]:
                dp[i][j] = 1 + dp[i + 1][j + 1]        # match -> diagonal + 1
            else:
                dp[i][j] = max(dp[i + 1][j], dp[i][j + 1])  # no match -> best of skipping one
    return dp[0][0]
````

````python
# Correctness check
tests = [("abcde","ace",3), ("abc","abc",3), ("abc","def",0), ("bl","yby",1)]
for a, b, exp in tests:
    x, y = lcs_brute(a, b), lcs_dp(a, b)
    print(f"{a!r},{b!r} -> brute={x}, dp={y} | expected={exp}")
    assert x == y == exp, "mismatch!"
print("\nAll tests passed")
````

## ⏱️ Empirically Checking the Complexities

Big-O can't be read off a function directly, but it can be **measured**. We time each approach on inputs of growing size `n` and read the **doubling ratio** — how much runtime grows when `n` doubles.

| Theoretical | Ratio when `n` → `2n` |
|-------------|-----------------------|
| `O(n)`        | ≈ **2×** |
| `O(n log n)`  | ≈ **2×** (slightly more) |
| `O(n²)`       | ≈ **4×** |

Inputs are shaped to force the worst case. (Exponential brute-force versions are shown in the code but omitted from timing where they would blow up — noted per notebook.)

*(Brute force is omitted from timing; we scale the DP with the string length.)*

````python
import os, sys
_root = os.getcwd()
for _ in range(5):
    if os.path.exists(os.path.join(_root, "bench_utils.py")):
        break
    _root = os.path.dirname(_root)
if _root not in sys.path:
    sys.path.insert(0, _root)
from bench_utils import benchmark

def make_worst_case(n):
    s = 'ab' * (n // 2 + 1)
    s = s[:n]
    return (s, s)   # identical strings -> full DP table work
solutions = {
    "dp O(m*n)": lcs_dp,
}
sizes = [200, 400, 800, 1600]

benchmark(solutions, make_worst_case, sizes, plot=True)
````

## 🧩 Patterns Learned

- **2-D grid DP over two sequences:** the template for edit distance, LCS, and string alignment.
- **Match / skip recurrence:** compare current items; take them if equal, else skip from the better side.
- **Signal:** "longest/shortest common ... of two strings", "edit distance", "align two sequences".
- **Related problems:** Edit Distance, Longest Common Substring, Distinct Subsequences.
- **Common pitfalls:** (1) off-by-one on table size (`+1`); (2) subsequence vs substring.
