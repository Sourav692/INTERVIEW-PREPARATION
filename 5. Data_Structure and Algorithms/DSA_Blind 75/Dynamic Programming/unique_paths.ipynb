# 62. Unique Paths
**Difficulty:** 🟡 Medium · **Topic:** Dynamic Programming · **LeetCode:** https://leetcode.com/problems/unique-paths/

## 💡 Concepts

**Core concept(s):** Grid DP — paths to a cell = paths from **above** + paths from the **left**.

**Why it applies here:** A robot moving only right or down reaches a cell only from the cell above or the cell to its left. So the count for each cell is the sum of those two. Fill the grid once; the bottom-right holds the answer. (A math shortcut using combinations also works.)

**Key intuition:** Ways to a cell = ways from the cell above + ways from the cell to the left.

---

### 📚 What is Dynamic Programming (DP)?
**DP** solves a big problem by solving smaller **overlapping** subproblems once and reusing the answers. Two styles: **memoization** (recursion that caches results) and **tabulation** (fill a table from the smallest cases up).
- **Why it's fast:** it turns exponential re-computation into a single sweep over the subproblems.
- **In Python:** a `dict`/list cache, or a `dp` list/2-D table.

### 📚 Subproblems & Recurrence
The heart of DP is a **recurrence**: the answer for a state written in terms of smaller states (e.g. `dp[i] = dp[i-1] + dp[i-2]`). Find the recurrence and the base cases, and the code writes itself.

---

**Prerequisite knowledge:**
- 2-D (or rolling 1-D) grid DP.
- The additive recurrence.

## 📝 Problem

A robot at the top-left of an `m × n` grid moves only right or down. How many unique paths to the bottom-right?

**Example**
```
m=3, n=7 -> 28
```

> Two approaches: exponential brute force and `O(m×n)` DP (rolling to `O(n)` space).

### Approach 1 — Brute Recursion (worst)

**Idea:** From each cell, sum the paths going down and going right. Recomputes cells exponentially.

**Time:** `O(2^(m+n))`. **Space:** `O(m+n)`.

````python
def unique_paths_brute(m, n):
    def dfs(r, c):                         # number of paths from (r,c) to the bottom-right
        if r == m - 1 or c == n - 1:
            return 1                       # on the last row/column there is only one way
        return dfs(r + 1, c) + dfs(r, c + 1)   # go down + go right
    return dfs(0, 0)
````

### Approach 2 — Rolling 1-D DP (optimal)

**Idea:** Keep one row of counts. Each cell becomes itself (paths from above) plus the cell to its left (paths from the left).

**Time:** `O(m×n)`. **Space:** `O(n)`.

````python
def unique_paths_dp(m, n):
    row = [1] * n                          # one row of counts; top row is all 1s
    for _ in range(1, m):                  # process each remaining row
        for c in range(1, n):
            row[c] += row[c - 1]           # paths here = from above (row[c]) + from left (row[c-1])
    return row[n - 1]                      # bottom-right cell holds the answer
````

````python
# Correctness check
tests = [(3,7,28),(3,2,3),(1,1,1),(3,3,6)]
for m, n, exp in tests:
    a, b = unique_paths_brute(m, n), unique_paths_dp(m, n)
    print(f"m={m}, n={n} -> brute={a}, dp={b} | expected={exp}")
    assert a == b == exp, "mismatch!"
print("\nAll tests passed")
````

## ⏱️ Empirically Checking the Complexities

Big-O can't be read off a function directly, but it can be **measured**. We time each approach on inputs of growing size `n` and read the **doubling ratio** — how much runtime grows when `n` doubles.

| Theoretical | Ratio when `n` → `2n` |
|-------------|-----------------------|
| `O(n)`        | ≈ **2×** |
| `O(n log n)`  | ≈ **2×** (slightly more) |
| `O(n²)`       | ≈ **4×** |

Inputs are shaped to force the worst case. (Exponential brute-force versions are shown in the code but omitted from timing where they would blow up — noted per notebook.)

*(Brute force explodes, so we time it on small square grids alongside the DP.)*

````python
import os, sys
_root = os.getcwd()
for _ in range(5):
    if os.path.exists(os.path.join(_root, "bench_utils.py")):
        break
    _root = os.path.dirname(_root)
if _root not in sys.path:
    sys.path.insert(0, _root)
from bench_utils import benchmark

def make_worst_case(n):
    return (n, n)   # n x n grid
solutions = {
    "brute O(2^(m+n))": unique_paths_brute,
    "dp    O(m*n)    ": unique_paths_dp,
}
sizes = [6, 8, 10, 12]

benchmark(solutions, make_worst_case, sizes, plot=True)
````

## 🧩 Patterns Learned

- **Grid DP with an additive recurrence:** each cell sums its two predecessors.
- **Roll a 2-D table into 1-D:** you only need the previous row.
- **Signal:** "count paths in a grid moving right/down", "ways to reach a cell".
- **Related problems:** Unique Paths II (obstacles), Minimum Path Sum, Dungeon Game.
- **Common pitfalls:** (1) exponential recursion without memo; (2) initializing the first row/column wrong.
