## 💡 Concepts

**Core concept(s):** **Sort by start**, then check neighbors for overlap.

**Why it applies here:** One person can attend all meetings only if none overlap. After sorting by start time, a conflict can only be between consecutive meetings — so one pass over neighbors settles it.

**Key intuition:** Sort by start; if any meeting starts before the previous one ends, there's a clash.

---

### 📚 Sorting to Reveal Order
Sorting intervals by their start (or end) puts overlapping ones next to each other, so a single left-to-right pass can merge or count them. Cost: **O(n log n)**.

---

**Prerequisite knowledge:**
- Sorting.
- Adjacent overlap check.

## 📝 Problem

Given meeting time intervals, can a person attend **all** of them (no overlaps)?

**Example**
```
[[0,30],[5,10],[15,20]] -> False
[[7,10],[2,4]] -> True
```

> One approach: sort by start, check consecutive pairs, `O(n log n)`.

### Approach — Sort + Adjacent Check

**Idea:** Sort by start. If any meeting begins before the previous ends, they conflict.

**Time:** `O(n log n)`. **Space:** `O(1)`.

````python
def can_attend_meetings(intervals):
    intervals.sort(key=lambda x: x[0])     # sort by start time
    for i in range(1, len(intervals)):
        if intervals[i][0] < intervals[i - 1][1]:   # this meeting starts before the previous ends
            return False                   # -> they overlap, can't attend both
    return True
````

````python
# Correctness check
tests = [([[0,30],[5,10],[15,20]],False), ([[7,10],[2,4]],True), ([],True), ([[1,2],[2,3]],True)]
for iv, exp in tests:
    got = can_attend_meetings([x[:] for x in iv])
    print(f"{iv} -> {got}")
    assert got == exp, "mismatch!"
print("\nAll tests passed")
````

| `O(n log n)` | ≈ **2×** (slightly more) |
| `O(n²)`     | ≈ **4×** |

## 🧩 Patterns Learned

- **Sort then check neighbors:** after sorting, conflicts are always adjacent.
- **Signal:** "can attend all / any overlap".
