### Valid Parentheses

```python
# 1. Create a dictionary like {")":"(","}":"{","]","["}
# 2. Iterate over the string 
#   1. If it is opening bracket push it to stack (dictionary contains the closing bracket as key so if we search any element then it would be the closing bracket only).
#   2. If it is closing bracket
#     1. Check if stack is non empty and its the value of the closing bracket is matching with that top element. 
#     2. if yes pop from Stack
#     3. Otherwise Return False

def isValid(s: str) -> bool:
    closeToOpen = {")":"(","}":"{","]":"["}
    stack = []

    for elem in s:
        if elem in closeToOpen:
            if stack and stack[-1] == closeToOpen[elem]:
                stack.pop()
            else:
                return False
        else:
            stack.append(elem)
    
    return True if not stack else False
```

```python
s = "[]"
res = isValid(s)
print(res)
```

```python
s = "([{}])"
res = isValid(s)
print(res)
```

### Baseball Game
You are keeping the scores for a baseball game with strange rules. At the beginning of the game, you start with an empty record.

Given a list of strings operations, where operations[i] is the ith operation you must apply to the record and is one of the following:

An integer x: Record a new score of x.

'+': Record a new score that is the sum of the previous two scores.

'D': Record a new score that is the double of the previous score.

'C': Invalidate the previous score, removing it from the record.

Return the sum of all the scores on the record after applying all the operations.

```python
def calPoints(operations: List[str]) -> int:
    stack = []

    for elem in operations:
        if elem.isnumeric():
            stack.append(int(elem))
        elif elem == "+":
            new_elem = stack[-1] + stack[-2]
            stack.append(new_elem)
        elif elem == "D":
            new_elem = stack[-1]*2
            stack.append(new_elem)
        else:
            stack.pop()
    return sum(stack)

```

```python
ops = ["1","2","+","C","5","D"]

res = calPoints(ops)
res
```
